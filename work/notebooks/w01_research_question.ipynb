## 1. My lane (or freestyle) and why

**Lane: Refresh / Content Opportunity Scoring.** I chose this lane because it turns search and content signals into a practical review queue. The goal is not to predict Google or guarantee that a page will recover. The goal is to identify which pages should be reviewed first and provide evidence for the review decision.

```python3
print("Provisional lane: Refresh / Content Opportunity Scoring")
print("Focus: prioritize pages for human review using observed search and content signals.")
```

## 2. The question: decision, action, cost of a wrong call

**Decision:** Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring?

**Who acts:** A content or SEO reviewer acts on the ranked recommendations.

**Action:** The reviewer uses the ranked queue and supporting signals to decide which pages deserve attention first and what type of review is appropriate.

**Cost of a wrong call:** A false positive can waste limited review time on a low-priority page. A false negative can cause a potentially important page to be reviewed too late. Because review capacity is limited, ranking the highest-priority pages first is useful.

```python3
print("Decision: prioritize pages for content review.")
print("Action: review the highest-priority pages first and choose an appropriate content action.")
print("Main costs: wasted reviewer time from false positives and missed opportunities from false negatives.")
```

## 4. Careful words: what I can and can't claim

This work can make **observed and measured** claims about patterns in the available data. It can provide **directional, decision-support** recommendations for prioritizing pages for human review.

It cannot prove that refreshing a page will cause traffic or rankings to recover. It cannot establish causal relationships from these observations alone, and it cannot predict Google's ranking decisions or claim to know Google's algorithm. Results will therefore be presented as evidence from the available data rather than guarantees.

```python3
allowed_claim_types = ["observed", "measured", "directional", "decision-support"]
disallowed_claims = ["causal proof", "guaranteed recovery", "predicting Google"]

print("Allowed claim types:", ", ".join(allowed_claim_types))
print("Claims explicitly avoided:", ", ".join(disallowed_claims))
```
